```python
import torch
import numpy as np
import matplotlib.pyplot as plt
from kymatio.torch import Scattering1D

log_eps = 1e-6
```

```python
x = np.arange(0, 100000, dtype=np.float32) / 1000.
y = np.sin(x).astype(np.float32) + \
    np.sin(1/2.*x).astype(np.float32) + \
    np.sin(1/4.*x).astype(np.float32) + \
    np.sin(1/8.*x).astype(np.float32) + \
    np.sin(1/16.*x).astype(np.float32) + \
    np.sin(12.*x).astype(np.float32) + \
    np.sin(24.*x).astype(np.float32) + \
    np.sin(48.*x).astype(np.float32) + \
    np.sin(96.*x).astype(np.float32) + \
    np.sin(192.*x).astype(np.float32)
```

```python
plt.plot(y)
```

```python
T = x.shape[-1]
J = 4
Q = 1
scattering = Scattering1D(J, T, Q)
sx = scattering.forward(torch.from_numpy(y))
sx = torch.log(torch.abs(sx) + log_eps)

print("SX Shape:", sx.shape)

# Order 0
plt.plot(sx[0, :])
plt.show()

# Order 1
plt.plot(sx[1, :])
plt.plot(sx[2, :])
plt.plot(sx[3, :])
plt.plot(sx[4, :])
plt.plot(sx[5, :])
plt.show()

# Order 2
plt.plot(sx[6, :])
plt.plot(sx[7, :])
plt.plot(sx[8, :])
plt.plot(sx[9, :])
plt.plot(sx[10, :])
plt.plot(sx[11, :])
plt.plot(sx[12, :])
plt.show()
```
